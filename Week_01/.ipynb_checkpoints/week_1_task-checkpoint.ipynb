The dataset contains 1,000 rows and 7 columns. The missing values are as follows: Applicant Name (52), Email (352), Phone (182), Domain Applied (0), University (43), Status (130), and Application Date (0). No duplicate rows were found, and all columns are currently stored as strings

```python
df=df.dropna(subset=['Applicant Name'])
```

```python
df['Email']=df['Email'].fillna('not_provide@gmail.com')
```

```python
df['University']=df['University'].fillna('Not Specified')
```

```python
df['Phone'] = df['Phone'].fillna('Unknown')
```

```python
df['Status']=df['Status'].fillna('Under Review')
```

```python
df['Domain Applied'].nunique()
```

```python
count=0
for value in df['Domain Applied'].unique():
    print(value)
    count+=1
print(count)
```

```python
df['Domain Applied']=(df['Domain Applied'].str.strip().str.lower())
```

```python
domain_mapping={
   'web dev': 'Web Development',
    'artificial intelligence': 'Artificial Intelligence',
    'ai': 'Artificial Intelligence',
    'ui/ux': 'UI/UX',
    'data science': 'Data Science',
    'data sci': 'Data Science',
    'mobile app': 'Mobile Application',
    'mobile dev': 'Mobile Application'
}
df['Domain Applied']=(df['Domain Applied'].map(domain_mapping))
```

```python
for value in df['Domain Applied'].unique():
    print(value)
```

```python
df['Domain Applied'].isna().sum()
```

```python
for stat in df['Status'].unique():
    print(stat)
```

```python
status_mapping={
    'selected':'Selected',
    'rejected': 'Rejected',
    'under review': 'Under Review'
    
}
df['Status']=(df['Status'].str.lower().str.strip())
```

```python
df['Status']=df['Status'].map(status_mapping)
```

```python
for stat in df['Status'].unique():
    print(stat)
```

```python
print(df.dtypes)
```

```python
df['Application Date']=pd.to_datetime(df['Application Date'].astype(str).str.strip(),format='mixed',errors='coerce')
```

```python
df['Application Date'].isna().sum()
```

```python
print(df.dtypes)
```

```python
df['Phone'] = df['Phone'].str.replace('-', '', regex=False).str.replace(' ',
'', regex=False)
```

```python
df.isna().sum()
```

```python
df.info()
```

```python
df.to_csv('applicants_cleaned.csv', index=False)
```

### Data Quality Summary

**Starting dataset:** 1000 rows, 7 columns[cite: 1]

**Issues found:**
- 353 missing values in Email, 170 in Phone, 123 in Status, 59 in University, and 50 in Applicant Name[cite: 1]
- Domain Applied had 528 different spelling and casing variations for only 5 real domains[cite: 1]
- Application Date was stored as text, not as an actual date type[cite: 1]
- Status had inconsistent casing and formatting (e.g., "SELECTED", " under review ")[cite: 1]
- Phone numbers contained inconsistent dash and space formatting[cite: 1]

**Actions taken:**
- Dropped 50 rows with missing Applicant Name (unusable without identification)[cite: 1]
- Filled missing Email with "not_provide@gmail.com"[cite: 1]
- Filled missing University with "Not Specified" (non-critical field)[cite: 1]
- Filled missing Phone with "Unknown"[cite: 1]
- Filled missing Status with "Under Review"[cite: 1]
- Standardized all Domain Applied values into 5 consistent categories[cite: 1]
- Standardized Status values into 3 consistent categories (Selected, Under Review, Rejected)[cite: 1]
- Converted Application Date to a proper datetime format[cite: 1]
- Cleaned Phone number formatting for consistency by removing dashes and spaces[cite: 1]
**Final dataset:** 950 rows, 7 columns, no missing values in critical fields, no duplicates, consistent formatting throughout[cite: 1].
